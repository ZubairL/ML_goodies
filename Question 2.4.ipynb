```python

sns.set_style("darkgrid")
sns.set_palette("dark")
plt.rcParams.update({"font.size": 12})
fig, axes = plt.subplots(2, 2, figsize=(12, 12))
b = 0.5
c = 2
subset = df[df['orientation'] == c]
sns.kdeplot(subset['acc'], bw_adjust=b, ax=axes[0, 0])  
axes[0, 0].set_title(f"KDE of Acc for Orientation = {c}")
axes[0, 0].set_xlabel("Acc Value")
axes[0, 0].set_ylabel("Density")
axes[0, 0].legend([f"KDE for orientation {c}"])

c = 3
subset = df[df['orientation'] == c]
sns.kdeplot(subset['acc'], bw_adjust=b, ax=axes[0, 1])  
axes[0, 1].set_title(f"KDE of Acc for Orientation = {c}")
axes[0, 1].set_xlabel("Acc Value")
axes[0, 1].set_ylabel("Density")
axes[0, 1].legend([f"KDE for orientation {c}"])

c = 4
subset = df[df['orientation'] == c]
sns.kdeplot(subset['acc'], bw_adjust=b, ax=axes[1, 0])  
axes[1, 0].set_title(f"KDE of Acc for Orientation = {c}")
axes[1, 0].set_xlabel("Acc Value")
axes[1, 0].set_ylabel("Density")
axes[1, 0].legend([f"KDE for orientation {c}"])

c = 5
subset = df[df['orientation'] == c]
sns.kdeplot(subset['acc'], bw_adjust=b, ax=axes[1, 1])  
axes[1, 1].set_title(f"KDE of Acc for Orientation = {c}")
axes[1, 1].set_xlabel("Acc Value")
axes[1, 1].set_ylabel("Density")
axes[1, 1].legend([f"KDE for orientation {c}"])

plt.tight_layout()
plt.show()


fig, axes = plt.subplots(2, 2, figsize=(12, 12))

c = 2
subset = df[df['orientation'] == c]
sns.kdeplot(subset['amygdala'], bw_adjust=b, ax=axes[0, 0])  
axes[0, 0].set_title(f"KDE of Amygdala for Orientation = {c}")
axes[0, 0].set_xlabel("Amygdala Value")
axes[0, 0].set_ylabel("Density")
axes[0, 0].legend([f"KDE for orientation {c}"])

c = 3
subset = df[df['orientation'] == c]
sns.kdeplot(subset['amygdala'], bw_adjust=b, ax=axes[0, 1])  
axes[0, 1].set_title(f"KDE of Amygdala for Orientation = {c}")
axes[0, 1].set_xlabel("Amygdala Value")
axes[0, 1].set_ylabel("Density")
axes[0, 1].legend([f"KDE for orientation {c}"])

c = 4
subset = df[df['orientation'] == c]
sns.kdeplot(subset['amygdala'], bw_adjust=b, ax=axes[1, 0])  
axes[1, 0].set_title(f"KDE of Amygdala for Orientation = {c}")
axes[1, 0].set_xlabel("Amygdala Value")
axes[1, 0].set_ylabel("Density")
axes[1, 0].legend([f"KDE for orientation {c}"])

c = 5
subset = df[df['orientation'] == c]
sns.kdeplot(subset['amygdala'], bw_adjust=b, ax=axes[1, 1])  
axes[1, 1].set_title(f"KDE of Amygdala for Orientation = {c}")
axes[1, 1].set_xlabel("Amygdala Value")
axes[1, 1].set_ylabel("Density")
axes[1, 1].legend([f"KDE for orientation {c}"])

plt.tight_layout()
plt.show()
```
