```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('data/n90pol.csv')
```

```python
print(df)
```

```python
"""Sources/Background research:
https://stackoverflow.com/questions/4150171/how-to-create-a-density-plot
https://stackoverflow.com/questions/33203645/how-to-plot-a-histogram-using-matplotlib-in-python-with-a-list-of-data

"""
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

df.hist('amygdala', bins=13, ax=axes[0])
axes[0].set_title("Amygdala Histogram")  
axes[0].set_xlabel("Amygdala Value")
axes[0].set_ylabel("Frequency")

df.hist('acc', bins=13, ax=axes[1])
axes[1].set_title("ACC Histogram")  
axes[1].set_xlabel("ACC Value")
axes[1].set_ylabel("Frequency")

plt.tight_layout()  


fig, axes = plt.subplots(1, 2, figsize=(16, 6))
# KDE plot for Amygdala
sns.kdeplot(df['amygdala'], bw_adjust=0.5, ax=axes[0], label="Amygdala KDE", color="blue")
axes[0].set_title("Kernel Density Plot - Amygdala")
axes[0].set_xlabel("Amygdala Value")
axes[0].set_ylabel("Density")
axes[0].legend()

# KDE plot for ACC
sns.kdeplot(df['acc'], bw_adjust=0.5, ax=axes[1], label="ACC KDE", color="green")
axes[1].set_title("Kernel Density Plot - ACC")
axes[1].set_xlabel("ACC Value")
axes[1].set_ylabel("Density")
axes[1].legend()
```
