```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('data/n90pol.csv')
```

```python
x = df['amygdala']
y = df['acc']

# Define parameters for the 2D histogram
nbin = 10  # Number of bins
xedges = np.linspace(x.min(), x.max(), nbin + 1)
yedges = np.linspace(y.min(), y.max(), nbin + 1)

# Create the 2D histogram
hist, xedges, yedges = np.histogram2d(x, y, bins=[xedges, yedges])

# Prepare the data for 3D plotting
xpos, ypos = np.meshgrid(xedges[:-1] + xedges[1] / 2, yedges[:-1] + yedges[1] / 2, indexing="ij")
xpos = xpos.flatten()
ypos = ypos.flatten()
zpos = np.zeros_like(xpos)

# Define the dimensions of the bars
dx = dy = np.diff(xedges)[0]  # Bin width
dz = hist.flatten()

# Plot the 3D histogram
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(xpos, ypos, zpos, dx, dy, dz, zsort='average', color='teal')

# Set labels and title
ax.set_title('2D Histogram - ACC vs. Amygdala', fontsize=18)
ax.set_xlabel('Amygdala', fontsize=14)
ax.set_ylabel('ACC', fontsize=14)
ax.set_zlabel('Counts', fontsize=14)

plt.show()
```
